## Techniques to Handle Null Values
So, how do we handle missing data? We’ve got a few options, and the best one depends on the type of missing data and the size of the gaps in our dataset.

1. ## Removing Null Values
We can simply remove the rows or columns with null values. This works when there are only a few missing entries and we don’t mind losing a bit of data.

## Imputation
Imputation is about filling in the missing values with estimates, making our dataset complete again. There are several ways to do this:

Mean/Median/Mode Imputation: Fill in the missing value with the average, median, or most common value in the column.

```python
df=pd.read_csv(r'C:\Users\Dhruv stark\Desktop\machine learning practice\market_data.csv')
df.info()
```

```python
#impute the null value with numerical value 

df['Item_Weight'].fillna(df['Item_Weight'].median(),inplace=True)
```

```python
df.info()
```

```python
#impute the null value with the mode to the object non numerical value 
df['Outlet_Size'].fillna(df['Outlet_Size'].mode()[0],inplace=True)
```

```python
df.info()
```

## K-Nearest Neighbors (KNN) Imputation: Use similar data points (neighbors) to predict the missing value.

```python
df=pd.read_csv(r'C:\Users\Dhruv stark\Desktop\machine learning practice\market_data.csv')
df.columns
```

```python
from sklearn.impute import KNNImputer 

## KNNImputer only works on numeric columns
# So separate numeric and categorical first

nurmeric_col =['Item_Weight','Item_Visibility', 'Item_MRP', 'Outlet_Establishment_Year', 'Item_Outlet_Sales']
imputer = KNNImputer(n_neighbors=5)
df[nurmeric_col] = imputer.fit_transform(df[nurmeric_col])

#caterigorial columns uses the mode method 
df.info()
```
