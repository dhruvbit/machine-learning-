## Analysis

- When we look at the values of Area, the curve are not symmetric.
- They are skewed - meaning most values are clustered in a range, but a few extreme values (outliers) stretch the distribution toward one side.

## Why this is a problem for ML?

- Many machine learning algorithms (like linear regression, logistic regression, KNN, neural networks) are sensitive to feature scaling.
- Outliers can distort scaling methods (like MinMaxScaler or StandardScaler) → making most data appear compressed into a small range.
- As a result, the model may give too much importance to these rare extreme values.

## Solution - using the clip function

- clip() in pandas limits values to within a given range
- By clipping extreme values, You reduce skewness in the distribution.
- Scaling methods (like StandardScaler/MinMaxScaler) now work more effectively.
- The model becomes more robust and doesn't overfit to rare outliers.

## BIVARIENT means comparing 1 or 2 ATTRIBUTES to target data

```python
#bivarient means comparing 1 or 2 data to target data

cat_label=[ 'area']
for col in cat_label:
    plt.figure(figsize=(13,4))
    sns.scatterplot(x=col,y="price",data=df)
    plt.title("bivarient plot scatter ",fontsize=11)    
    plt.xlabel(col,fontsize=11)
    plt.ylabel("price",fontsize=11)
    plt.show
```

## why we use scatterplot and boxplot 
scatter plot because we have to compare the 2 numerical value 
boxplot because we have to compare the two category data 

```python
#boxplot 

cat_label=[ 'area', 'bedrooms', 'bathrooms', 'stories', 'mainroad',
       'guestroom', 'basement', 'hotwaterheating', 'airconditioning',
       'parking', 'prefarea', 'furnishingstatus']
for col in cat_label:
    plt.figure(figsize=(13,4))
    sns.boxplot(x=col,y="price",data=df)
    plt.title("bivarient plot scatter ",fontsize=11)    
    plt.xlabel(col,fontsize=11)
    plt.ylabel("price",fontsize=11)   #middle line called median of graph
    plt.show
```

## MULTIVARIENT means comparing MORE THAN 1 OR 2 ATTRIBUTE data to target ATTRIBUTE DATA

```python
df.info()
```

```python
multicol=['price', 'area', 'bedrooms', 'bathrooms', 'stories',  'parking']
plt.figure(figsize=(10,8))

#Highly correlated features (|r| > 0.9) are often redundant — they carry the same information. 
correlation_matrix = df[multicol].astype(float).corr()
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm',vmin=-1,vmax=1)
plt.title('heatmap')
plt.show()
```

```python
multicol = ['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking']
plt.figure(figsize=(10,8))

correlation_matrix = df[multicol].astype(float).corr()
sns.heatmap(correlation_matrix,
            fmt='.2f',
            cmap='coolwarm', 
            vmin=-1, vmax=1,
            linewidths=0.5, 
            linecolor='white', 
            annot_kws={"size": 14})
plt.title('Correlation Heatmap', fontsize=16, pad=15)
plt.tight_layout()
plt.show()
```
