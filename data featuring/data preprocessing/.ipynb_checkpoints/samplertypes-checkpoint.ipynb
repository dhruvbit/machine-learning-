## CLASS IMBALANCE 

What is Class Imbalance?
Class imbalance occurs when one class has significantly more samples than another in your training dataset.
Real world examples:

Fraud detection → 99% normal transactions, 1% fraudulent
Medical diagnosis → 95% healthy patients, 5% diseased


Why is it a Problem?
The model develops a majority class bias — it learns that predicting the dominant class almost always yields high accuracy, without actually learning meaningful patterns from the minority class.
Example:
Fraud dataset: 99% normal, 1% fraud

Trivial classifier predicts "Not Fraud" for everything:
Accuracy = 99%  ← statistically high
Recall on fraud = 0%  ← never catches a single fraud case

High accuracy here is completely misleading.
The accuracy metric becomes statistically dishonest under class imbalance. You must use Precision, Recall, and F1-Score instead.

How to Handle Class Imbalance
Data-Level Methods — fix the data distribution
Oversampling → increase minority class samples → best method is SMOTE which generates synthetic samples between existing minority class points rather than duplicating them.
Undersampling → reduce majority class samples → Random Undersampling removes majority rows until both classes are equal — risk is losing valuable information.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Logistic Regression
model = LogisticRegression(class_weight='balanced', max_iter=1000)

# Random Forest
model = RandomForestClassifier(class_weight='balanced', random_state=42)
What balanced does internally
Normal training:
Wrong on class0 → penalized 1x
Wrong on class1 → penalized 1x

With class_weight='balanced':
Wrong on class0 (99%) → penalized 1x
Wrong on class1 (1%)  → penalized 99x

Model is forced to pay attention to minority class
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
from imblearn.under_sampling import RandomUnderSampler,EditedNearestNeighbours
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from imblearn.over_sampling import RandomOverSampler,SMOTE,ADASYN

df = pd.read_csv(r'C:\Users\Dhruv stark\Desktop\machine learning practice\creditcard.csv')
df.info()
```

```python
df.describe().T
```

```python
df.head()
```

```python
plt.figure(figsize=(11,8))
sns.countplot(x=df["Class"])
plt.title('class dist')
plt.show()
```

```python
x = df.drop(['Class','Time'],axis = 1)
y = df['Class']

x.head()

x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=42,test_size=0.2,stratify=y)
```

```python
#baseline model logistic regression 

model = LogisticRegression(max_iter=1000,random_state=42)
model.fit(x_train,y_train)

y_predict = model.predict(x_test)
accuracy_score(y_test,y_predict)
```

```python
# RANDOMOVERSAMPLER
```

```python
ros = RandomOverSampler(random_state=42)
x_resampler,y_resampler=ros.fit_resample(x_train,y_train)

y_count = y_resampler.value_counts()
y_count
```

```python
plt.figure(figsize=(12,6))
plt.bar(y_count.index,y_count.values,color=['orange','blue'])
plt.xlabel('count')
plt.ylabel('value')
plt.show()
```

```python
model.fit(x_resampler,y_resampler)
```

```python
y_predict_ros = model.predict(x_test)
```

```python
accuracy_score(y_test,y_predict_ros)
```

"With replacement" means the same original row can be duplicated many times.

Step 3 — Duplicate those rows exactly
python# Original minority row:
Row 4: Age=28, Spending=500, HomePlanet='Mars', Fraud=1

# Gets copied and added as:
Row 1001: Age=28, Spending=500, HomePlanet='Mars', Fraud=1  ← exact copy
Row 1002: Age=28, Spending=500, HomePlanet='Mars', Fraud=1  ← exact copy
Row 1003: Age=28, Spending=500, HomePlanet='Mars', Fraud=1  ← exact copy
Zero intelligence. Pure copy-paste.

Step 4 — Final dataset
Original 900 class0 rows     → untouched
Original 100 class1 rows     → untouched
800 duplicated class1 rows   → added

Final: 900 class0, 900 class1 = balanced

Why This is Problematic
Model sees Row 4 copied 8 times:

Iteration 1 → sees Row 4 → learns pattern
Iteration 2 → sees Row 4 again → reinforces same pattern
Iteration 3 → sees Row 4 again → overfits to this exact row
...

Model memorizes specific rows
Instead of learning general patterns
This is called OVERFITTING

How SMOTE Fixes This
Instead of copying — SMOTE creates brand new rows that never existed:
Original minority rows:
Row A: Age=25, Spending=100
Row B: Age=30, Spending=150

SMOTE picks two neighbors A and B
Creates new point somewhere between them:

New Row: Age=27.5, Spending=125  ← never existed, not a copy

Formula:
new_point = row_A + random(0,1) × (row_B - row_A)
          = 25 + 0.5 × (30-25) = 27.5 for Age
          = 100 + 0.5 × (150-100) = 125 for Spending

Side by Side Comparison
RandomOverSampler:          SMOTE:
Row 4 copied 8 times        8 new synthetic rows created
↓                           ↓
Model memorizes Row 4       Model learns general minority pattern
↓                           ↓
Overfitting                 Better generalization



```python
#SMOTE

smote =SMOTE(random_state=42)
x_resm,y_resm = smote.fit_resample(x_train,y_train)

y_counts = y_resm.value_counts()
y_counts

```

```python
model.fit(x_resm,y_resm)


```

```python
y_predict_smote = model.predict(x_test)

accuracy_score(y_test, y_predict_smote)
```
