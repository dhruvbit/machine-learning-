## What is categorical data?
Some columns in your dataset contain labels or names instead of numbers — like "Male/Female" or "Delhi/Mumbai/Kolkata". These are called categorical features. There's no math you can do between them — "Delhi + Mumbai" means nothing.
The problem is that ML models only understand numbers. So before you feed this data into a model, you have to convert those labels into numbers. That conversion process is called encoding.

Two types of categorical data:
1. Nominal (unordered)
The categories have no ranking — none is "greater than" or "less than" another.
Example: Colors — Red, Blue, Green. Blue isn't more than Red.
2. Ordinal (ordered)
The categories do have a meaningful ranking, but the gaps between them aren't fixed numbers.
Example: Satisfaction — Poor < Average < Good < Excellent. "Excellent" is clearly better than "Poor", but you can't say it's exactly 3x better.

Why does this distinction matter?
How you encode the data depends on which type it is. If you blindly assign numbers (1, 2, 3) to nominal data like cities, the model will falsely assume Mumbai (2) is "between" Delhi (1) and Kolkata (3) — which is nonsense.
