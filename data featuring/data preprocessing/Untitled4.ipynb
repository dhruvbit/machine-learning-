```python
import numpy as np 
# positive label distribution node  
parents_node = 10 /(10+6)
child_1 = 8/(8+2)
child_2 = 2/(4+2)


#entropy 
h_parents = -parents_node *np.log2(parents_node) -( 1-parents_node)*np.log2(1-parents_node)
h_parents
pchild_1 =-child_1*np.log2(child_1)-(1-child_1)*np.log2(1-child_1) 
pchild_2 =-child_2*np.log2(child_2)-(1-child_2)*np.log2(1-child_2)  
#ratio of the parent and child  
s = (8+2)/(10+6)
final_child1 = (s*pchild_1) +((1-s)*pchild_2)

#information gain 

info_gain = h_parents - final_child1

info_gain

```
